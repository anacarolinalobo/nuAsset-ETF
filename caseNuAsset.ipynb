```python
# dado historico de volume da B3
# 2008 a 2026

import pandas as pd
from pathlib import Path


def ler_cotahist(path):

    names = [
        "tipo_registro",
        "data",
        "cod_bdi",
        "ticker",
        "tipo_mercado",
        "nome_empresa",
        "especificacao",
        "prazo_termo",
        "moeda",
        "preco_abertura",
        "preco_maximo",
        "preco_minimo",
        "preco_medio",
        "preco_ultimo",
        "preco_oferta_compra",
        "preco_oferta_venda",
        "numero_negocios",
        "quantidade_negociada",
        "volume_financeiro",
        "preco_exercicio",
        "indicador_correcao",
        "data_vencimento",
        "fator_cotacao",
        "pontos_exercicio",
        "isin",
        "numero_distribuicao"
    ]

    widths = [
        2, 8, 2, 12, 3, 12, 10, 3, 4,
        13, 13, 13, 13, 13, 13, 13,
        5, 18, 18, 13, 1, 8, 7, 13, 12, 3
    ]

    df = pd.read_fwf(
        path,
        widths=widths,
        names=names,
        dtype=str,
        header=None
    )

    # Apenas registros de cotação
    df = df[df["tipo_registro"] == "01"].copy()

    # Limpeza
    for col in ["ticker", "nome_empresa", "especificacao", "moeda", "isin"]:
        df[col] = df[col].fillna("").str.strip()

    # Data
    df["data"] = pd.to_datetime(
        df["data"],
        format="%Y%m%d",
        errors="coerce"
    )

    # Preços
    preco_cols = [
        "preco_abertura",
        "preco_maximo",
        "preco_minimo",
        "preco_medio",
        "preco_ultimo",
        "preco_oferta_compra",
        "preco_oferta_venda"
    ]

    for col in preco_cols:
        df[col] = (
            pd.to_numeric(df[col], errors="coerce") / 100
        )

    # Número de negócios
    df["numero_negocios"] = pd.to_numeric(
        df["numero_negocios"],
        errors="coerce"
    )

    # Quantidade de ações negociadas
    df["quantidade_negociada"] = pd.to_numeric(
        df["quantidade_negociada"],
        errors="coerce"
    )

    # Volume financeiro em R$
    df["volume_financeiro"] = (
        pd.to_numeric(
            df["volume_financeiro"],
            errors="coerce"
        ) / 100
    )

    # Mercado à vista
    df = df[df["tipo_mercado"] == "010"].copy()

    # Remove registros inválidos
    df = df[
        df["ticker"].ne("") &
        df["data"].notna() &
        df["preco_ultimo"].notna()
    ].copy()

    return df
```

```python
import pandas as pd
from pathlib import Path


# ============================================================
# 1. CONFIGURAÇÃO
# ============================================================

PASTA_B3 = Path("./")

# Layout do COTAHIST
COLSPECS = [
    (0, 2),      # tipo_registro
    (2, 10),     # data
    (10, 12),    # cod_bdi
    (12, 24),    # ticker
    (24, 27),    # tipo_mercado
    (27, 39),    # nome_empresa
    (39, 49),    # especificacao
    (49, 52),    # prazo_termo
    (52, 56),    # moeda
    (56, 69),    # preco_abertura
    (69, 82),    # preco_maximo
    (82, 95),    # preco_minimo
    (95, 108),   # preco_medio
    (108, 121),  # preco_ultimo
    (121, 134),  # preco_oferta_compra
    (134, 147),  # preco_oferta_venda
    (147, 152),  # numero_negocios
    (152, 170),  # quantidade_negociada
    (170, 188),  # volume_financeiro
    (188, 201),  # preco_exercicio
    (201, 202),  # indicador_correcao
    (202, 210),  # data_vencimento
    (210, 217),  # fator_cotacao
    (217, 230),  # pontos_exercicio
    (230, 242),  # isin
    (242, 245),  # numero_distribuicao
]

NOMES = [
    "tipo_registro",
    "data",
    "cod_bdi",
    "ticker",
    "tipo_mercado",
    "nome_empresa",
    "especificacao",
    "prazo_termo",
    "moeda",
    "preco_abertura",
    "preco_maximo",
    "preco_minimo",
    "preco_medio",
    "preco_ultimo",
    "preco_oferta_compra",
    "preco_oferta_venda",
    "numero_negocios",
    "quantidade_negociada",
    "volume_financeiro",
    "preco_exercicio",
    "indicador_correcao",
    "data_vencimento",
    "fator_cotacao",
    "pontos_exercicio",
    "isin",
    "numero_distribuicao",
]


# ============================================================
# 2. FUNÇÃO PARA LER UM ARQUIVO
# ============================================================

def ler_cotahist(arquivo):

    df = pd.read_fwf(
        arquivo,
        colspecs=COLSPECS,
        names=NOMES,
        encoding="latin1",
        dtype=str
    )

    # --------------------------------------------------------
    # Mantém somente registros de negociação
    # --------------------------------------------------------
    df = df[df["tipo_registro"] == "01"].copy()

    # --------------------------------------------------------
    # Data
    # --------------------------------------------------------
    df["data"] = pd.to_datetime(
        df["data"],
        format="%Y%m%d",
        errors="coerce"
    )

    # --------------------------------------------------------
    # Ticker
    # --------------------------------------------------------
    df["ticker"] = df["ticker"].str.strip()

    # --------------------------------------------------------
    # Campos numéricos
    # --------------------------------------------------------

    colunas_numericas = [
        "preco_abertura",
        "preco_maximo",
        "preco_minimo",
        "preco_medio",
        "preco_ultimo",
        "preco_oferta_compra",
        "preco_oferta_venda",
        "numero_negocios",
        "quantidade_negociada",
        "volume_financeiro",
        "preco_exercicio",
        "fator_cotacao",
        "pontos_exercicio"
    ]

    for coluna in colunas_numericas:
        df[coluna] = pd.to_numeric(
            df[coluna],
            errors="coerce"
        )

    # --------------------------------------------------------
    # Ajustes dos campos monetários
    # --------------------------------------------------------

    colunas_preco = [
        "preco_abertura",
        "preco_maximo",
        "preco_minimo",
        "preco_medio",
        "preco_ultimo",
        "preco_oferta_compra",
        "preco_oferta_venda",
        "preco_exercicio"
    ]

    for coluna in colunas_preco:
        df[coluna] = df[coluna] / 100

    # Volume
    df["volume_financeiro"] = df["volume_financeiro"] / 100

    return df


# ============================================================
# 3. ENCONTRAR TODOS OS ARQUIVOS
# ============================================================

arquivos = sorted(
    PASTA_B3.glob("COTAHIST_A*.TXT")
)

print(f"Arquivos encontrados: {len(arquivos)}")

for arquivo in arquivos:
    print(arquivo.name)


# ============================================================
# 4. LER TODOS OS ANOS
# ============================================================

lista_dfs = []

for arquivo in arquivos:

    print(f"Lendo: {arquivo.name}")

    try:

        df_ano = ler_cotahist(arquivo)

        # Guarda o ano de origem
        df_ano["ano_arquivo"] = arquivo.stem[-4:]

        lista_dfs.append(df_ano)

        print(
            f"   {len(df_ano):,} registros"
        )

    except Exception as e:

        print(
            f"ERRO em {arquivo.name}: {e}"
        )


# ============================================================
# 5. JUNTAR TODOS OS ANOS
# ============================================================

df_b3 = pd.concat(
    lista_dfs,
    ignore_index=True
)


# ============================================================
# 6. ORDENAR
# ============================================================

df_b3 = df_b3.sort_values(
    ["data", "ticker"]
).reset_index(drop=True)


# ============================================================
# 7. RESULTADO
# ============================================================

print("\n==============================")
print("DATAFRAME FINAL")
print("==============================")

print(f"Linhas: {len(df_b3):,}")
print(f"Colunas: {len(df_b3.columns)}")
print(
    f"Período: "
    f"{df_b3['data'].min()} "
    f"até "
    f"{df_b3['data'].max()}"
)

print(
    f"Tickers únicos: "
    f"{df_b3['ticker'].nunique():,}"
)


# ============================================================
# 8. SALVAR
# ============================================================

df_b3.to_parquet(
    "b3_cotahist_2008_2026.parquet",
    index=False
)

print("\nArquivo salvo:")
print("b3_cotahist_2008_2026.parquet")
```

```python
import pandas as pd
import os
import glob

# ============================================================
# 1. CONFIGURAÇÃO
# ============================================================

anos = range(2010, 2027)

arquivos = []

# ============================================================
# 2. ENCONTRAR SOMENTE OS ARQUIVOS DE CAPITAL SOCIAL
# ============================================================

for ano in anos:

    pasta = f"fre_cia_aberta_{ano}"

    padrao = os.path.join(
        pasta,
        f"fre_cia_aberta_capital_social_{ano}.csv"
    )

    encontrados = glob.glob(padrao)

    arquivos.extend(encontrados)

print(f"Total de arquivos encontrados: {len(arquivos)}")

for arquivo in arquivos:
    print(arquivo)


# ============================================================
# 3. FUNÇÃO PARA LER CSV COM DIFERENTES ENCODINGS
# ============================================================

def ler_csv_robusto(arquivo):

    encodings = [
        "utf-8",
        "latin1",
        "cp1252"
    ]

    separadores = [
        ";",
        ","
    ]

    ultimo_erro = None

    for encoding in encodings:

        for sep in separadores:

            try:

                df = pd.read_csv(
                    arquivo,
                    sep=sep,
                    encoding=encoding,
                    low_memory=False,
                    on_bad_lines="skip"
                )

                # Evita aceitar uma leitura claramente errada
                # (ex.: CSV separado por ; lido como uma única coluna)
                if len(df.columns) > 1:

                    print(
                        f"OK: {arquivo} | "
                        f"encoding={encoding} | "
                        f"sep='{sep}' | "
                        f"linhas={len(df)} | "
                        f"colunas={len(df.columns)}"
                    )

                    return df

            except Exception as e:

                ultimo_erro = e

    raise ValueError(
        f"Não foi possível ler o arquivo: {arquivo}\n"
        f"Último erro: {ultimo_erro}"
    )


# ============================================================
# 4. LER TODOS OS ARQUIVOS
# ============================================================

dfs = []

for arquivo in arquivos:

    try:

        df_temp = ler_csv_robusto(arquivo)

        # Guarda o ano de origem
        ano = os.path.basename(arquivo)[-8:-4]

        df_temp["ano_arquivo"] = int(ano)

        # Guarda o arquivo de origem
        df_temp["arquivo_origem"] = os.path.basename(arquivo)

        dfs.append(df_temp)

    except Exception as e:

        print(f"ERRO: {arquivo}")
        print(e)


# ============================================================
# 5. JUNTAR TODOS OS DATAFRAMES
# ============================================================

if dfs:

    df_capital_social = pd.concat(
        dfs,
        ignore_index=True,
        sort=False
    )

else:

    df_capital_social = pd.DataFrame()


# ============================================================
# 6. LIMPAR NOMES DAS COLUNAS
# ============================================================

df_capital_social.columns = (
    df_capital_social.columns
    .str.strip()
    .str.replace("\ufeff", "", regex=False)
)


# ============================================================
# 7. REMOVER ESPAÇOS DAS COLUNAS DE TEXTO
# ============================================================

for coluna in df_capital_social.select_dtypes(
    include="object"
).columns:

    df_capital_social[coluna] = (
        df_capital_social[coluna]
        .str.strip()
    )


# ============================================================
# 8. INFORMAÇÕES FINAIS
# ============================================================

print("\n===================================")
print("DATAFRAME FINAL")
print("===================================")

print("Shape:", df_capital_social.shape)

print("\nColunas:")
print(df_capital_social.columns.tolist())

print("\nPrimeiras linhas:")
display(df_capital_social.head())

print("\nQuantidade por ano:")
print(
    df_capital_social["ano_arquivo"]
    .value_counts()
    .sort_index()
)
```

```python
 import pandas as pd
import numpy as np
import requests
import io
import re
import unicodedata


# ============================================================
# 1. BAIXAR CADASTRO DA CVM
# ============================================================

url_cvm = (
    "https://dados.cvm.gov.br/"
    "dados/cia_aberta/CAD/DADOS/cad_cia_aberta.csv"
)

response = requests.get(url_cvm)

response.raise_for_status()

cad_cvm = pd.read_csv(
    io.BytesIO(response.content),
    sep=";",
    encoding="latin1",
    dtype=str
)

print("Cadastro CVM:")
print(cad_cvm.shape)
print(cad_cvm.columns.tolist())


# ============================================================
# 2. NORMALIZAR CNPJ
# ============================================================

def normalizar_cnpj(x):

    if pd.isna(x):
        return np.nan

    x = str(x)

    # Remove tudo que não for número
    x = re.sub(r"\D", "", x)

    if len(x) == 0:
        return np.nan

    # CNPJ deve ter 14 dígitos
    return x.zfill(14)


# CVM
cad_cvm["CNPJ"] = (
    cad_cvm["CNPJ_CIA"]
    .apply(normalizar_cnpj)
)


# Capital social
cs = df_capital_social.copy()

cs["CNPJ"] = (
    cs["CNPJ_Companhia"]
    .apply(normalizar_cnpj)
)


# ============================================================
# 3. PREPARAR CADASTRO CVM
# ============================================================

cad_cvm = cad_cvm[
    [
        "CNPJ",
        "DENOM_SOCIAL",
        "DENOM_COMERC",
        "SIT"
    ]
].copy()

cad_cvm = cad_cvm.drop_duplicates(
    subset=["CNPJ"]
)


# ============================================================
# 4. CRUZAR CAPITAL SOCIAL COM CVM
# ============================================================

cs = cs.merge(
    cad_cvm,
    on="CNPJ",
    how="left",
    suffixes=("", "_CVM")
)


print("\nCapital social após merge com CVM:")

print(
    cs[
        [
            "CNPJ",
            "Nome_Companhia",
            "DENOM_SOCIAL"
        ]
    ].head()
)


# ============================================================
# 5. CONVERTER DATA
# ============================================================

cs["Data_Referencia"] = pd.to_datetime(
    cs["Data_Referencia"],
    errors="coerce"
)

cs["mes"] = (
    cs["Data_Referencia"]
    .dt.to_period("M")
)


# ============================================================
# 6. QUANTIDADE DE AÇÕES
# ============================================================

cs["Quantidade_Total_Acoes"] = pd.to_numeric(
    cs["Quantidade_Total_Acoes"],
    errors="coerce"
)

cs["Quantidade_Acoes_Ordinarias"] = pd.to_numeric(
    cs["Quantidade_Acoes_Ordinarias"],
    errors="coerce"
)

cs["Quantidade_Acoes_Preferenciais"] = pd.to_numeric(
    cs["Quantidade_Acoes_Preferenciais"],
    errors="coerce"
)

cs = cs.dropna(
    subset=[
        "CNPJ",
        "mes",
        "Quantidade_Total_Acoes"
    ]
)


# ============================================================
# 7. PEGAR A INFORMAÇÃO MAIS RECENTE DO CAPITAL
#    DENTRO DE CADA MÊS
# ============================================================

cs = cs.sort_values(
    [
        "CNPJ",
        "mes",
        "Data_Referencia"
    ]
)


capital_mensal = (
    cs
    .groupby(
        [
            "CNPJ",
            "mes"
        ],
        as_index=False
    )
    .last()
)


# ============================================================
# 8. PREPARAR DF_B3
# ============================================================

b3 = df_b3.copy()

b3["data"] = pd.to_datetime(
    b3["data"],
    errors="coerce"
)

b3["ticker"] = (
    b3["ticker"]
    .astype(str)
    .str.strip()
    .str.upper()
)

b3["preco_ultimo"] = pd.to_numeric(
    b3["preco_ultimo"],
    errors="coerce"
)

b3 = b3.dropna(
    subset=[
        "data",
        "ticker",
        "preco_ultimo"
    ]
)


# ============================================================
# 9. MÊS DA COTAÇÃO
# ============================================================

b3["mes"] = (
    b3["data"]
    .dt.to_period("M")
)


# ============================================================
# 10. PEGAR ÚLTIMO PREÇO DE CADA TICKER/MÊS
# ============================================================

b3_mensal = (
    b3
    .sort_values(
        [
            "ticker",
            "data"
        ]
    )
    .groupby(
        [
            "ticker",
            "mes"
        ],
        as_index=False
    )
    .last()
)


# ============================================================
# 11. NORMALIZAR NOMES
# ============================================================

def normalizar_nome(x):

    if pd.isna(x):
        return ""

    x = str(x).upper().strip()

    # Remove acentos
    x = unicodedata.normalize(
        "NFKD",
        x
    )

    x = (
        x
        .encode("ASCII", "ignore")
        .decode("ASCII")
    )

    # Remove pontuação
    x = re.sub(
        r"[^A-Z0-9\s]",
        " ",
        x
    )

    # Remove espaços duplicados
    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


# ============================================================
# 12. NOME DA EMPRESA NO CAPITAL SOCIAL
# ============================================================

cs["empresa_norm"] = (
    cs["Nome_Companhia"]
    .apply(normalizar_nome)
)


# ============================================================
# 13. NOME DA EMPRESA NA B3
# ============================================================

b3_mensal["empresa_norm"] = (
    b3_mensal["nome_empresa"]
    .apply(normalizar_nome)
)


# ============================================================
# 14. CRIAR MAPA EMPRESA → TICKER
# ============================================================

mapa_empresa_ticker = (
    b3_mensal[
        [
            "empresa_norm",
            "ticker",
            "isin"
        ]
    ]
    .drop_duplicates()
)


# ============================================================
# 15. CRUZAR CAPITAL SOCIAL COM TICKER
# ============================================================

cs_ticker = cs.merge(
    mapa_empresa_ticker,
    on="empresa_norm",
    how="left"
)


# ============================================================
# 16. CRUZAR COM PREÇOS MENSAIS
# ============================================================

df_market_cap = cs_ticker.merge(
    b3_mensal[
        [
            "ticker",
            "mes",
            "data",
            "preco_ultimo",
            "isin"
        ]
    ],
    on=[
        "ticker",
        "mes"
    ],
    how="left",
    suffixes=("", "_b3")
)


# ============================================================
# 17. CALCULAR MARKET CAP
# ============================================================

df_market_cap["market_cap"] = (
    df_market_cap["preco_ultimo"]
    *
    df_market_cap["Quantidade_Total_Acoes"]
)


# ============================================================
# 18. SELECIONAR COLUNAS FINAIS
# ============================================================

df_market_cap = df_market_cap[
    [
        "mes",
        "data",
        "ticker",
        "isin",
        "CNPJ",
        "Nome_Companhia",
        "DENOM_SOCIAL",
        "Quantidade_Total_Acoes",
        "preco_ultimo",
        "market_cap"
    ]
]


# ============================================================
# 19. ORDENAR
# ============================================================

df_market_cap = (
    df_market_cap
    .sort_values(
        [
            "mes",
            "ticker"
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# 20. RESULTADO
# ============================================================

print("\n========================================")
print("RESULTADO")
print("========================================")

print(
    "Shape:",
    df_market_cap.shape
)

print(
    "\nColunas:",
    df_market_cap.columns.tolist()
)

display(
    df_market_cap.head(20)
)
```

```python
import pandas as pd
import numpy as np
import re
import unicodedata


# ============================================================
# CONFIGURAÇÕES
# ============================================================

ANO_INICIAL = 2010
ANO_FINAL = 2026


# ============================================================
# 1. FUNÇÃO DE NORMALIZAÇÃO DE NOME
# ============================================================

def normalizar_nome(x):

    if pd.isna(x):
        return ""

    x = str(x).upper().strip()

    x = unicodedata.normalize("NFKD", x)

    x = "".join(
        c for c in x
        if not unicodedata.combining(c)
    )

    x = re.sub(
        r"[^A-Z0-9 ]",
        " ",
        x
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    )

    return x.strip()


# ============================================================
# 2. PREPARAR CAPITAL SOCIAL
# ============================================================

cs = df_capital_social.copy()


# CNPJ
cs["CNPJ"] = (
    cs["CNPJ_Companhia"]
    .astype(str)
    .str.replace(r"\D", "", regex=True)
    .str.zfill(14)
)


# Data
cs["Data_Referencia"] = pd.to_datetime(
    cs["Data_Referencia"],
    errors="coerce"
)

cs["mes"] = (
    cs["Data_Referencia"]
    .dt.to_period("M")
)


# Quantidades
colunas_acoes = [
    "Quantidade_Acoes_Ordinarias",
    "Quantidade_Acoes_Preferenciais",
    "Quantidade_Total_Acoes"
]

for col in colunas_acoes:

    cs[col] = pd.to_numeric(
        cs[col],
        errors="coerce"
    )


# Remover registros inválidos
cs = cs.dropna(
    subset=[
        "CNPJ",
        "Data_Referencia",
        "mes"
    ]
)


# ============================================================
# 3. PEGAR ÚLTIMO REGISTRO DE CAPITAL DE CADA MÊS
# ============================================================

cs = cs.sort_values(
    [
        "CNPJ",
        "mes",
        "Data_Referencia"
    ]
)

cs = (
    cs
    .groupby(
        ["CNPJ", "mes"],
        as_index=False
    )
    .last()
)


# ============================================================
# 4. PREPARAR B3
# ============================================================

b3 = df_b3.copy()


b3["data"] = pd.to_datetime(
    b3["data"],
    errors="coerce"
)

b3["mes"] = (
    b3["data"]
    .dt.to_period("M")
)


b3["ticker"] = (
    b3["ticker"]
    .astype(str)
    .str.upper()
    .str.strip()
)


b3["preco_ultimo"] = pd.to_numeric(
    b3["preco_ultimo"],
    errors="coerce"
)

b3["volume_financeiro"] = pd.to_numeric(
    b3["volume_financeiro"],
    errors="coerce"
)

b3["numero_negocios"] = pd.to_numeric(
    b3["numero_negocios"],
    errors="coerce"
)

b3["quantidade_negociada"] = pd.to_numeric(
    b3["quantidade_negociada"],
    errors="coerce"
)


# ============================================================
# 5. FILTRAR MERCADO À VISTA
# ============================================================

b3["tipo_mercado"] = (
    b3["tipo_mercado"]
    .astype(str)
    .str.strip()
)


b3 = b3[
    b3["tipo_mercado"] == "010"
].copy()


# ============================================================
# 6. FILTRAR SOMENTE AÇÕES
# ============================================================

# 3 = ON
# 4 = PN
# 5-8 = classes de PN

b3 = b3[
    b3["ticker"].str.contains(
        r"[345678]$",
        regex=True,
        na=False
    )
].copy()


# ============================================================
# 7. IDENTIFICAR ON / PN
# ============================================================

def identificar_tipo_acao(ticker):

    ultimo = str(ticker)[-1]

    if ultimo == "3":
        return "ON"

    if ultimo in ["4", "5", "6", "7", "8"]:
        return "PN"

    return np.nan


b3["tipo_acao"] = (
    b3["ticker"]
    .apply(identificar_tipo_acao)
)


# ============================================================
# 8. REMOVER PREÇOS INVÁLIDOS
# ============================================================

b3 = b3[
    b3["preco_ultimo"].notna()
    &
    (b3["preco_ultimo"] > 0)
].copy()


# ============================================================
# 9. PREÇO DE FIM DE MÊS
# ============================================================

b3 = b3.sort_values(
    [
        "ticker",
        "data"
    ]
)


preco_mensal = (
    b3
    .groupby(
        ["ticker", "mes"],
        as_index=False
    )
    .last()
)


preco_mensal = preco_mensal[
    [
        "ticker",
        "mes",
        "data",
        "preco_ultimo",
        "isin",
        "tipo_acao"
    ]
].copy()


# ============================================================
# 10. LIQUIDEZ MENSAL
# ============================================================

liquidez_mensal = (
    b3
    .groupby(
        ["ticker", "mes"],
        as_index=False
    )
    .agg(

        volume_financeiro=(
            "volume_financeiro",
            "sum"
        ),

        volume_financeiro_medio=(
            "volume_financeiro",
            "mean"
        ),

        numero_negocios=(
            "numero_negocios",
            "sum"
        ),

        quantidade_negociada=(
            "quantidade_negociada",
            "sum"
        ),

        dias_negociados=(
            "data",
            "nunique"
        )
    )
)


# ============================================================
# 11. CRUZAR PREÇO + LIQUIDEZ
# ============================================================

universo = preco_mensal.merge(
    liquidez_mensal,
    on=[
        "ticker",
        "mes"
    ],
    how="left"
)


# ============================================================
# 12. PREPARAR NOMES PARA O MATCH CVM × B3
# ============================================================

cs["nome_companhia_norm"] = (
    cs["Nome_Companhia"]
    .apply(normalizar_nome)
)


# Cadastro CVM
cadastro_nomes = cs[
    [
        "CNPJ",
        "nome_companhia_norm"
    ]
].copy()


cadastro_nomes = cadastro_nomes.rename(
    columns={
        "nome_companhia_norm": "nome_b3"
    }
)


# ============================================================
# 13. NOMES DA B3
# ============================================================

b3["nome_empresa_norm"] = (
    b3["nome_empresa"]
    .apply(normalizar_nome)
)


mapa_b3 = b3[
    [
        "ticker",
        "nome_empresa_norm",
        "tipo_acao",
        "isin"
    ]
].copy()


mapa_b3 = mapa_b3.rename(
    columns={
        "nome_empresa_norm": "nome_b3"
    }
)


mapa_b3 = mapa_b3[
    mapa_b3["nome_b3"] != ""
].drop_duplicates()


# ============================================================
# 14. MAPEAR CNPJ → TICKER
# ============================================================

mapa_ticker = cadastro_nomes.merge(
    mapa_b3,
    on="nome_b3",
    how="inner"
)


mapa_ticker = mapa_ticker[
    [
        "CNPJ",
        "ticker",
        "tipo_acao",
        "isin"
    ]
].drop_duplicates()


# ============================================================
# 15. CRUZAR CAPITAL SOCIAL COM TICKER
# ============================================================

capital_ticker = cs.merge(
    mapa_ticker,
    on="CNPJ",
    how="inner",
    suffixes=(
        "_capital",
        "_b3"
    )
)


# ============================================================
# 16. ESCOLHER QUANTIDADE DA ESPÉCIE
# ============================================================

capital_ticker["Quantidade_Acoes"] = np.where(

    capital_ticker["tipo_acao"] == "ON",

    capital_ticker[
        "Quantidade_Acoes_Ordinarias"
    ],

    np.where(

        capital_ticker["tipo_acao"] == "PN",

        capital_ticker[
            "Quantidade_Acoes_Preferenciais"
        ],

        np.nan
    )
)


# ============================================================
# 17. REMOVER QUANTIDADES INVÁLIDAS
# ============================================================

capital_ticker = capital_ticker[
    capital_ticker["Quantidade_Acoes"].notna()
    &
    (
        capital_ticker["Quantidade_Acoes"] > 0
    )
].copy()


# ============================================================
# 18. EVITAR DUPLICIDADE CNPJ/TICKER/MÊS
# ============================================================

capital_ticker = (
    capital_ticker
    .sort_values(
        [
            "CNPJ",
            "ticker",
            "mes",
            "Data_Referencia"
        ]
    )
    .drop_duplicates(
        subset=[
            "CNPJ",
            "ticker",
            "mes"
        ],
        keep="last"
    )
)


# ============================================================
# 19. CRUZAR COM PREÇO
# ============================================================

df_universo = capital_ticker.merge(
    universo,
    on=[
        "ticker",
        "mes"
    ],
    how="inner",
    suffixes=(
        "_capital",
        ""
    )
)


# ============================================================
# 20. MARKET CAP
# ============================================================

df_universo["market_cap"] = (
    df_universo["Quantidade_Acoes"]
    *
    df_universo["preco_ultimo"]
)


# ============================================================
# 21. ORGANIZAR POR TICKER/MÊS
# ============================================================

df_universo = df_universo.sort_values(
    [
        "ticker",
        "mes"
    ]
).reset_index(drop=True)


# ============================================================
# 22. PREÇO 12 MESES ATRÁS
# ============================================================

df_universo["preco_12m_atras"] = (
    df_universo
    .groupby("ticker")["preco_ultimo"]
    .shift(12)
)


# ============================================================
# 23. RETORNO 12 MESES
# ============================================================

df_universo["retorno_12m"] = (
    df_universo["preco_ultimo"]
    /
    df_universo["preco_12m_atras"]
    - 1
)


# ============================================================
# 24. MOMENTUM RANK
# ============================================================

df_universo["momentum_rank"] = (
    df_universo
    .groupby("mes")["retorno_12m"]
    .rank(
        ascending=False,
        method="first"
    )
)


# ============================================================
# 25. MOMENTUM PERCENTILE
# ============================================================

df_universo["momentum_percentile"] = (
    df_universo
    .groupby("mes")["retorno_12m"]
    .rank(
        ascending=True,
        pct=True
    )
)


# ============================================================
# 26. RANK MARKET CAP
# ============================================================

df_universo["market_cap_rank"] = (
    df_universo
    .groupby("mes")["market_cap"]
    .rank(
        ascending=False,
        method="first"
    )
)


# ============================================================
# 27. RANK LIQUIDEZ
# ============================================================

df_universo["liquidez_rank"] = (
    df_universo
    .groupby("mes")["volume_financeiro_medio"]
    .rank(
        ascending=False,
        method="first"
    )
)


# ============================================================
# 28. FILTRAR PERÍODO
# ============================================================

df_universo = df_universo[
    (
        df_universo["mes"]
        >= pd.Period(
            f"{ANO_INICIAL}-01",
            freq="M"
        )
    )
    &
    (
        df_universo["mes"]
        <= pd.Period(
            f"{ANO_FINAL}-12",
            freq="M"
        )
    )
].copy()


# ============================================================
# 29. EXIGIR 12 MESES DE HISTÓRICO
# ============================================================

df_universo = df_universo[
    df_universo["retorno_12m"].notna()
].copy()


# ============================================================
# 30. SELECIONAR COLUNAS FINAIS
# ============================================================

colunas_finais = [
    "mes",
    "data",
    "ticker",
    "tipo_acao",
    "isin",

    "CNPJ",
    "Nome_Companhia",

    "Quantidade_Acoes",
    "Quantidade_Acoes_Ordinarias",
    "Quantidade_Acoes_Preferenciais",
    "Quantidade_Total_Acoes",

    "preco_ultimo",
    "preco_12m_atras",

    "market_cap",

    "volume_financeiro",
    "volume_financeiro_medio",

    "numero_negocios",
    "quantidade_negociada",
    "dias_negociados",

    "retorno_12m",

    "momentum_rank",
    "momentum_percentile",

    "market_cap_rank",
    "liquidez_rank"
]


# só manter as que realmente existem
colunas_finais = [
    c for c in colunas_finais
    if c in df_universo.columns
]


df_universo = df_universo[
    colunas_finais
].copy()


# ============================================================
# 31. ORDENAR
# ============================================================

df_universo = df_universo.sort_values(
    [
        "mes",
        "momentum_rank"
    ]
).reset_index(drop=True)


# ============================================================
# 32. DIAGNÓSTICO
# ============================================================

print("=" * 60)
print("DF UNIVERSO")
print("=" * 60)

print(
    "Shape:",
    df_universo.shape
)

print(
    "Período:",
    df_universo["mes"].min(),
    "até",
    df_universo["mes"].max()
)

print(
    "Tickers únicos:",
    df_universo["ticker"].nunique()
)

print(
    "Empresas únicas:",
    df_universo["CNPJ"].nunique()
)

print("\nTipos de ação:")
print(
    df_universo["tipo_acao"]
    .value_counts()
)

print("\nExemplo:")
print(
    df_universo.head(20)
)
```

```python
# ============================================================
# MAPEAMENTO B3 -> CVM -> CNPJ
# ============================================================

import pandas as pd
import numpy as np
import re
import unicodedata

from rapidfuzz import process, fuzz


# ============================================================
# 1. NORMALIZAÇÃO DOS NOMES
# ============================================================

def normalizar_nome(nome):

    if pd.isna(nome):
        return ""

    nome = str(nome).upper().strip()

    # Remove acentos
    nome = unicodedata.normalize("NFKD", nome)
    nome = "".join(
        c for c in nome
        if not unicodedata.combining(c)
    )

    # Padroniza &
    nome = nome.replace("&", " E ")

    # Remove pontuação
    nome = re.sub(r"[^A-Z0-9\s]", " ", nome)

    # Remove tipos societários
    tipos_societarios = (
        r"\b(?:SA|S A|S/A|SAA|LTDA|LTD|PLC|INC|CORP|"
        r"CORPORATION|HOLDING|HOLDINGS)\b"
    )

    nome = re.sub(
        tipos_societarios,
        " ",
        nome
    )

    # Remove espaços duplicados
    nome = re.sub(r"\s+", " ", nome).strip()

    return nome


# ============================================================
# 2. PREPARAR UNIVERSO B3
# ============================================================

b3 = df_instrumentos_acoes.copy()

b3["ticker"] = (
    b3["ticker"]
    .astype(str)
    .str.strip()
    .str.upper()
)

b3["isin"] = (
    b3["isin"]
    .astype(str)
    .str.strip()
    .str.upper()
)

b3["nome_norm"] = (
    b3["nome_empresa"]
    .apply(normalizar_nome)
)

# Apenas tickers padrão:
# 4 letras + número de 3 a 8
b3 = b3[
    b3["ticker"].str.fullmatch(
        r"[A-Z]{4}[3-8]",
        na=False
    )
].copy()

# Apenas ON e PN
b3 = b3[
    b3["especificacao"]
    .astype(str)
    .str.upper()
    .str.split()
    .str[0]
    .isin(["ON", "PN"])
].copy()

# Remove registros sem informações necessárias
b3 = b3.dropna(
    subset=[
        "ticker",
        "isin",
        "nome_norm"
    ]
).copy()

# Uma observação por ticker + ISIN
b3 = (
    b3
    .sort_values("data")
    .drop_duplicates(
        subset=["ticker", "isin"],
        keep="last"
    )
    .reset_index(drop=True)
)

print("=" * 70)
print("UNIVERSO B3")
print("=" * 70)
print("Linhas:", len(b3))
print("Tickers:", b3["ticker"].nunique())
print("ISINs:", b3["isin"].nunique())


# ============================================================
# 3. PREPARAR CADASTRO CVM
# ============================================================

cvm = df_capital_social.copy()

cvm["CNPJ_Companhia"] = (
    cvm["CNPJ_Companhia"]
    .astype(str)
    .str.strip()
)

cvm["nome_norm"] = (
    cvm["Nome_Companhia"]
    .apply(normalizar_nome)
)

cvm = cvm.dropna(
    subset=[
        "CNPJ_Companhia",
        "nome_norm"
    ]
).copy()


# ============================================================
# 4. UMA LINHA POR EMPRESA DA CVM
# ============================================================

cvm_empresas = (
    cvm[
        [
            "CNPJ_Companhia",
            "Nome_Companhia",
            "nome_norm"
        ]
    ]
    .drop_duplicates(
        subset=["CNPJ_Companhia"]
    )
    .reset_index(drop=True)
)

print()
print("=" * 70)
print("CADASTRO CVM")
print("=" * 70)
print("Empresas:", len(cvm_empresas))
print(
    "CNPJs:",
    cvm_empresas["CNPJ_Companhia"].nunique()
)


# ============================================================
# 5. LISTA DE NOMES CVM
# ============================================================

nomes_cvm = (
    cvm_empresas["nome_norm"]
    .tolist()
)


# ============================================================
# 6. FUZZY MATCH
# ============================================================

resultados = []

for _, row in b3.iterrows():

    nome_b3 = row["nome_norm"]

    matches = process.extract(
        nome_b3,
        nomes_cvm,
        scorer=fuzz.token_set_ratio,
        limit=3
    )

    for nome_match, score, idx in matches:

        empresa_cvm = cvm_empresas.iloc[idx]

        resultados.append({
            "ticker": row["ticker"],
            "isin": row["isin"],
            "nome_b3": row["nome_empresa"],
            "nome_cvm": empresa_cvm["Nome_Companhia"],
            "CNPJ": empresa_cvm["CNPJ_Companhia"],
            "score": score
        })


df_matches = pd.DataFrame(resultados)


# ============================================================
# 7. CLASSIFICAR QUALIDADE
# ============================================================

df_matches["qualidade"] = np.select(
    [
        df_matches["score"] >= 95,
        df_matches["score"] >= 85,
        df_matches["score"] >= 75
    ],
    [
        "ALTO",
        "MEDIO",
        "BAIXO"
    ],
    default="REVISAR"
)


# ============================================================
# 8. MELHOR MATCH POR TICKER
# ============================================================

mapping_preliminar = (
    df_matches
    .sort_values(
        ["ticker", "score"],
        ascending=[True, False]
    )
    .drop_duplicates(
        subset=["ticker"]
    )
    .reset_index(drop=True)
)


# ============================================================
# 9. MATCHES ACEITOS
# ============================================================

mapping_validado = (
    mapping_preliminar[
        mapping_preliminar["score"] >= 85
    ]
    .copy()
    .reset_index(drop=True)
)


# ============================================================
# 10. MATCHES PARA REVISÃO
# ============================================================

mapping_revisar = (
    mapping_preliminar[
        mapping_preliminar["score"] < 85
    ]
    .copy()
    .reset_index(drop=True)
)


# ============================================================
# 11. ESTATÍSTICAS
# ============================================================

print()
print("=" * 70)
print("RESULTADO DO FUZZY MATCH")
print("=" * 70)

print(
    "Tickers B3:",
    len(mapping_preliminar)
)

print(
    "Score >= 95:",
    (mapping_preliminar["score"] >= 95).sum()
)

print(
    "Score 85-94:",
    (
        (mapping_preliminar["score"] >= 85)
        &
        (mapping_preliminar["score"] < 95)
    ).sum()
)

print(
    "Score 75-84:",
    (
        (mapping_preliminar["score"] >= 75)
        &
        (mapping_preliminar["score"] < 85)
    ).sum()
)

print(
    "Score < 75:",
    (mapping_preliminar["score"] < 75).sum()
)


# ============================================================
# 12. DISTRIBUIÇÃO DA QUALIDADE
# ============================================================

print()
print("QUALIDADE:")
print(
    mapping_preliminar["qualidade"]
    .value_counts()
)


# ============================================================
# 13. MELHORES MATCHES
# ============================================================

print()
print("=" * 70)
print("PRIMEIROS MATCHES")
print("=" * 70)

print(
    mapping_preliminar[
        [
            "ticker",
            "isin",
            "nome_b3",
            "nome_cvm",
            "CNPJ",
            "score",
            "qualidade"
        ]
    ]
    .sort_values(
        "score",
        ascending=False
    )
    .head(30)
    .to_string(index=False)
)


# ============================================================
# 14. MATCHES PROBLEMÁTICOS
# ============================================================

print()
print("=" * 70)
print("MATCHES PARA REVISÃO")
print("=" * 70)

print(
    mapping_revisar[
        [
            "ticker",
            "isin",
            "nome_b3",
            "nome_cvm",
            "CNPJ",
            "score",
            "qualidade"
        ]
    ]
    .sort_values(
        "score",
        ascending=True
    )
    .head(50)
    .to_string(index=False)
)


# ============================================================
# 15. TOP 3 CANDIDATOS DOS MATCHES DUVIDOSOS
# ============================================================

tickers_revisar = (
    mapping_revisar["ticker"]
    .tolist()
)

df_candidatos_revisao = (
    df_matches[
        df_matches["ticker"]
        .isin(tickers_revisar)
    ]
    .sort_values(
        ["ticker", "score"],
        ascending=[True, False]
    )
)

print()
print("=" * 70)
print("TOP 3 CANDIDATOS DOS MATCHES DUVIDOSOS")
print("=" * 70)

print(
    df_candidatos_revisao[
        [
            "ticker",
            "isin",
            "nome_b3",
            "nome_cvm",
            "CNPJ",
            "score"
        ]
    ]
    .head(100)
    .to_string(index=False)
)


# ============================================================
# 16. MAPPING FINAL
# ============================================================

mapping_validado = (
    mapping_validado[
        [
            "ticker",
            "isin",
            "nome_b3",
            "nome_cvm",
            "CNPJ",
            "score",
            "qualidade"
        ]
    ]
    .drop_duplicates(
        subset=["ticker"]
    )
    .reset_index(drop=True)
)


# ============================================================
# 17. RESUMO
# ============================================================

print()
print("=" * 70)
print("RESUMO FINAL")
print("=" * 70)

print(
    f"Universo B3:          {b3['ticker'].nunique():,}"
)

print(
    f"Mapping total:         {len(mapping_preliminar):,}"
)

print(
    f"Mapping validado:      {len(mapping_validado):,}"
)

print(
    f"Necessitam revisão:    {len(mapping_revisar):,}"
)

print(
    f"CNPJs únicos mapeados: "
    f"{mapping_validado['CNPJ'].nunique():,}"
)
```

```python
# ============================================================
# MATCHING B3 -> CVM MELHORADO
# ============================================================

import pandas as pd
import numpy as np
import re
import unicodedata

from rapidfuzz import process, fuzz


# ============================================================
# 1. NORMALIZAÇÃO
# ============================================================

def normalizar_nome(nome):

    if pd.isna(nome):
        return ""

    nome = str(nome).upper().strip()

    # Acentos
    nome = unicodedata.normalize("NFKD", nome)
    nome = "".join(
        c for c in nome
        if not unicodedata.combining(c)
    )

    # Padronização
    nome = nome.replace("&", " E ")

    # Pontuação
    nome = re.sub(r"[^A-Z0-9\s]", " ", nome)

    # Tipos societários
    nome = re.sub(
        r"\b(?:SA|S A|S/A|SAA|LTDA|LTD|PLC|INC|CORP|"
        r"CORPORATION|HOLDING|HOLDINGS)\b",
        " ",
        nome
    )

    # Espaços
    nome = re.sub(r"\s+", " ", nome).strip()

    return nome


# ============================================================
# 2. PREPARAR B3
# ============================================================

b3 = df_instrumentos_acoes.copy()

b3["ticker"] = (
    b3["ticker"]
    .astype(str)
    .str.strip()
    .str.upper()
)

b3["isin"] = (
    b3["isin"]
    .astype(str)
    .str.strip()
    .str.upper()
)

b3["nome_norm"] = (
    b3["nome_empresa"]
    .apply(normalizar_nome)
)

# Somente tickers padrão
b3 = b3[
    b3["ticker"].str.fullmatch(
        r"[A-Z]{4}[3-8]",
        na=False
    )
].copy()

# Somente ON / PN
b3 = b3[
    b3["especificacao"]
    .astype(str)
    .str.upper()
    .str.split()
    .str[0]
    .isin(["ON", "PN"])
].copy()

# Um registro por ticker
b3 = (
    b3
    .sort_values("data")
    .drop_duplicates("ticker")
    .reset_index(drop=True)
)


# ============================================================
# 3. PREPARAR CVM
# ============================================================

cvm = df_capital_social.copy()

cvm["CNPJ_Companhia"] = (
    cvm["CNPJ_Companhia"]
    .astype(str)
    .str.strip()
)

cvm["nome_norm"] = (
    cvm["Nome_Companhia"]
    .apply(normalizar_nome)
)

cvm_empresas = (
    cvm[
        [
            "CNPJ_Companhia",
            "Nome_Companhia",
            "nome_norm"
        ]
    ]
    .drop_duplicates("CNPJ_Companhia")
    .reset_index(drop=True)
)


# ============================================================
# 4. DICIONÁRIO DE NOMES CVM
# ============================================================

nomes_cvm = cvm_empresas["nome_norm"].tolist()


# ============================================================
# 5. MATCHING COM VÁRIOS ALGORITMOS
# ============================================================

resultados = []

for _, row in b3.iterrows():

    nome_b3 = row["nome_norm"]

    # Candidatos usando vários métodos
    candidatos = set()

    for scorer in [
        fuzz.ratio,
        fuzz.partial_ratio,
        fuzz.token_sort_ratio,
        fuzz.token_set_ratio,
        fuzz.WRatio
    ]:

        matches = process.extract(
            nome_b3,
            nomes_cvm,
            scorer=scorer,
            limit=5
        )

        for _, _, idx in matches:
            candidatos.add(idx)

    # Avaliar todos os candidatos
    for idx in candidatos:

        nome_cvm = cvm_empresas.iloc[idx]["nome_norm"]

        score_ratio = fuzz.ratio(
            nome_b3,
            nome_cvm
        )

        score_partial = fuzz.partial_ratio(
            nome_b3,
            nome_cvm
        )

        score_token_sort = fuzz.token_sort_ratio(
            nome_b3,
            nome_cvm
        )

        score_token_set = fuzz.token_set_ratio(
            nome_b3,
            nome_cvm
        )

        score_wratio = fuzz.WRatio(
            nome_b3,
            nome_cvm
        )

        # Score combinado
        score_final = (
            0.15 * score_ratio
            + 0.30 * score_partial
            + 0.15 * score_token_sort
            + 0.20 * score_token_set
            + 0.20 * score_wratio
        )

        resultados.append({
            "ticker": row["ticker"],
            "isin": row["isin"],
            "nome_b3": row["nome_empresa"],
            "nome_cvm": cvm_empresas.iloc[idx]["Nome_Companhia"],
            "CNPJ": cvm_empresas.iloc[idx]["CNPJ_Companhia"],
            "score_ratio": score_ratio,
            "score_partial": score_partial,
            "score_token_sort": score_token_sort,
            "score_token_set": score_token_set,
            "score_wratio": score_wratio,
            "score_final": score_final
        })


df_matches_melhorado = pd.DataFrame(resultados)


# ============================================================
# 6. MELHOR MATCH POR TICKER
# ============================================================

mapping_preliminar = (
    df_matches_melhorado
    .sort_values(
        ["ticker", "score_final"],
        ascending=[True, False]
    )
    .drop_duplicates("ticker")
    .reset_index(drop=True)
)


# ============================================================
# 7. DIFERENÇA ENTRE PRIMEIRO E SEGUNDO CANDIDATO
# ============================================================

ranking = (
    df_matches_melhorado
    .sort_values(
        ["ticker", "score_final"],
        ascending=[True, False]
    )
    .copy()
)

ranking["rank"] = (
    ranking
    .groupby("ticker")
    .cumcount() + 1
)

melhor = (
    ranking[ranking["rank"] == 1]
    [["ticker", "score_final"]]
    .rename(columns={
        "score_final": "score_1"
    })
)

segundo = (
    ranking[ranking["rank"] == 2]
    [["ticker", "score_final"]]
    .rename(columns={
        "score_final": "score_2"
    })
)

mapping_preliminar = mapping_preliminar.merge(
    melhor,
    on="ticker",
    how="left"
)

mapping_preliminar = mapping_preliminar.merge(
    segundo,
    on="ticker",
    how="left"
)

mapping_preliminar["diferenca_score"] = (
    mapping_preliminar["score_1"]
    - mapping_preliminar["score_2"]
)


# ============================================================
# 8. CLASSIFICAÇÃO
# ============================================================

mapping_preliminar["qualidade"] = np.select(
    [
        (
            (mapping_preliminar["score_1"] >= 90)
            &
            (mapping_preliminar["diferenca_score"] >= 10)
        ),

        (
            (mapping_preliminar["score_1"] >= 85)
            &
            (mapping_preliminar["diferenca_score"] >= 7)
        ),

        (
            (mapping_preliminar["score_1"] >= 80)
            &
            (mapping_preliminar["diferenca_score"] >= 5)
        )
    ],
    [
        "ALTO",
        "MEDIO",
        "BAIXO"
    ],
    default="REVISAR"
)


# ============================================================
# 9. SEPARAR RESULTADOS
# ============================================================

mapping_validado = (
    mapping_preliminar[
        mapping_preliminar["qualidade"].isin(
            ["ALTO", "MEDIO"]
        )
    ]
    .copy()
    .reset_index(drop=True)
)

mapping_revisar = (
    mapping_preliminar[
        mapping_preliminar["qualidade"] == "REVISAR"
    ]
    .copy()
    .reset_index(drop=True)
)


# ============================================================
# 10. RESULTADOS
# ============================================================

print("=" * 70)
print("MATCHING MELHORADO")
print("=" * 70)

print(
    "Total de tickers:",
    len(mapping_preliminar)
)

print(
    "ALTO:",
    (mapping_preliminar["qualidade"] == "ALTO").sum()
)

print(
    "MEDIO:",
    (mapping_preliminar["qualidade"] == "MEDIO").sum()
)

print(
    "BAIXO:",
    (mapping_preliminar["qualidade"] == "BAIXO").sum()
)

print(
    "REVISAR:",
    (mapping_preliminar["qualidade"] == "REVISAR").sum()
)


# ============================================================
# 11. MELHORES MATCHES
# ============================================================

print()
print("=" * 70)
print("MELHORES MATCHES")
print("=" * 70)

print(
    mapping_preliminar[
        [
            "ticker",
            "isin",
            "nome_b3",
            "nome_cvm",
            "CNPJ",
            "score_1",
            "score_2",
            "diferenca_score",
            "qualidade"
        ]
    ]
    .sort_values(
        "score_1",
        ascending=False
    )
    .head(30)
    .to_string(index=False)
)


# ============================================================
# 12. MATCHES QUE PRECISAM DE REVISÃO
# ============================================================

print()
print("=" * 70)
print("MATCHES PARA REVISÃO")
print("=" * 70)

print(
    mapping_revisar[
        [
            "ticker",
            "isin",
            "nome_b3",
            "nome_cvm",
            "CNPJ",
            "score_1",
            "score_2",
            "diferenca_score"
        ]
    ]
    .sort_values(
        "score_1",
        ascending=True
    )
    .head(50)
    .to_string(index=False)
)


# ============================================================
# 13. SALVAR RESULTADOS
# ============================================================

mapping_preliminar.to_csv(
    "mapping_b3_cvm_completo.csv",
    index=False
)

mapping_validado.to_csv(
    "mapping_b3_cvm_validado.csv",
    index=False
)

mapping_revisar.to_csv(
    "mapping_b3_cvm_revisar.csv",
    index=False
)

print()
print("=" * 70)
print("ARQUIVOS GERADOS")
print("=" * 70)
print("mapping_b3_cvm_completo.csv")
print("mapping_b3_cvm_validado.csv")
print("mapping_b3_cvm_revisar.csv")
```

```python
import pandas as pd
import numpy as np
import re
import unicodedata
from difflib import SequenceMatcher


# ============================================================
# CONFIGURAÇÃO
# ============================================================

ARQUIVO_SAIDA = "isin_cnpj_ticker_empresa.csv"

# Se quiser exigir um score mínimo para fuzzy matching
FUZZY_THRESHOLD = 0.85


# ============================================================
# 1. NORMALIZAÇÃO
# ============================================================

def normalizar_texto(x):

    if pd.isna(x):
        return ""

    x = str(x).upper().strip()

    # Remove acentos
    x = unicodedata.normalize("NFKD", x)
    x = "".join(
        c for c in x
        if not unicodedata.combining(c)
    )

    # Substituições
    x = x.replace("&", " E ")

    # Remove pontuação
    x = re.sub(r"[^A-Z0-9\s]", " ", x)

    # Remove termos societários comuns
    termos = [
        " S A ",
        " SA ",
        " S A",
        " CIA ",
        " COMPANHIA ",
        " CORPORATION ",
        " CORP ",
        " HOLDING ",
        " HOLDINGS ",
        " PARTICIPACOES ",
        " PARTICIPACAO ",
        " INVESTIMENTOS ",
        " FINANCEIRA ",
        " BANCO ",
    ]

    for termo in termos:
        x = x.replace(termo, " ")

    # Espaços duplicados
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def normalizar_cnpj(x):

    if pd.isna(x):
        return None

    x = re.sub(r"\D", "", str(x))

    if len(x) == 14:
        return x

    return None


def normalizar_ticker(x):

    if pd.isna(x):
        return ""

    return (
        str(x)
        .upper()
        .strip()
        .replace(" ", "")
    )


def normalizar_isin(x):

    if pd.isna(x):
        return ""

    return (
        str(x)
        .upper()
        .strip()
        .replace(" ", "")
    )


# ============================================================
# 2. PREPARAR df_b3
# ============================================================

b3 = df_b3.copy()

b3.columns = (
    b3.columns
    .str.strip()
    .str.lower()
)


# Verificar colunas esperadas

colunas_necessarias_b3 = [
    "isin",
    "ticker",
    "nome_empresa"
]

faltantes = [
    c for c in colunas_necessarias_b3
    if c not in b3.columns
]

if faltantes:

    raise ValueError(
        f"Colunas ausentes no df_b3: {faltantes}"
    )


# Normalização

b3["isin"] = b3["isin"].apply(normalizar_isin)

b3["ticker_norm"] = (
    b3["ticker"]
    .apply(normalizar_ticker)
)

b3["nome_norm"] = (
    b3["nome_empresa"]
    .apply(normalizar_texto)
)


# ============================================================
# 3. PREPARAR df_fca
# ============================================================

fca = df_fca.copy()

fca.columns = (
    fca.columns
    .str.strip()
)


colunas_necessarias_fca = [
    "CNPJ_Companhia",
    "Nome_Companhia"
]

faltantes = [
    c for c in colunas_necessarias_fca
    if c not in fca.columns
]

if faltantes:

    raise ValueError(
        f"Colunas ausentes no df_fca: {faltantes}"
    )


# CNPJ

fca["cnpj"] = (
    fca["CNPJ_Companhia"]
    .apply(normalizar_cnpj)
)


# Nome

fca["nome_norm"] = (
    fca["Nome_Companhia"]
    .apply(normalizar_texto)
)


# ============================================================
# 4. TENTAR ENCONTRAR TICKER NA CVM
# ============================================================

# Dependendo do seu arquivo FCA, pode não existir ticker.
# Procuramos automaticamente uma coluna que possa conter ticker.

possiveis_tickers = [
    c for c in fca.columns
    if "ticker" in c.lower()
    or "codigo_negociacao" in c.lower()
    or "codigo_negociacao" in c.lower()
]

if possiveis_tickers:

    coluna_ticker_cvm = possiveis_tickers[0]

    fca["ticker_norm"] = (
        fca[coluna_ticker_cvm]
        .apply(normalizar_ticker)
    )

else:

    fca["ticker_norm"] = ""


# ============================================================
# 5. REDUZIR df_fca PARA EMPRESAS ÚNICAS
# ============================================================

empresas = (
    fca[
        [
            "cnpj",
            "Nome_Companhia",
            "nome_norm",
            "ticker_norm"
        ]
    ]
    .dropna(subset=["cnpj"])
    .drop_duplicates()
)


# ============================================================
# 6. MATCH POR CNPJ
# ============================================================

# Se df_b3 tiver CNPJ, usa diretamente.

possiveis_cnpj_b3 = [
    c for c in b3.columns
    if "cnpj" in c.lower()
]

if possiveis_cnpj_b3:

    coluna_cnpj_b3 = possiveis_cnpj_b3[0]

    b3["cnpj"] = (
        b3[coluna_cnpj_b3]
        .apply(normalizar_cnpj)
    )

else:

    b3["cnpj"] = None


# ============================================================
# 7. MAPA CNPJ → EMPRESA
# ============================================================

mapa_cnpj = (
    empresas[
        [
            "cnpj",
            "Nome_Companhia"
        ]
    ]
    .drop_duplicates("cnpj")
)


b3 = b3.merge(
    mapa_cnpj,
    on="cnpj",
    how="left"
)


# ============================================================
# 8. MATCH POR TICKER
# ============================================================

# Criamos mapa ticker → CNPJ

ticker_cnpj = (
    empresas[
        [
            "ticker_norm",
            "cnpj"
        ]
    ]
    .query("ticker_norm != ''")
    .drop_duplicates("ticker_norm")
)


# Só preencher CNPJ ainda vazio

b3 = b3.merge(
    ticker_cnpj.rename(
        columns={
            "cnpj": "cnpj_ticker"
        }
    ),
    on="ticker_norm",
    how="left"
)


b3["cnpj"] = (
    b3["cnpj"]
    .fillna(b3["cnpj_ticker"])
)


b3.drop(
    columns=["cnpj_ticker"],
    inplace=True
)


# ============================================================
# 9. MATCH POR NOME EXATO
# ============================================================

nome_cnpj = (
    empresas[
        [
            "nome_norm",
            "cnpj"
        ]
    ]
    .query("nome_norm != ''")
    .drop_duplicates("nome_norm")
)


b3 = b3.merge(
    nome_cnpj.rename(
        columns={
            "cnpj": "cnpj_nome"
        }
    ),
    on="nome_norm",
    how="left"
)


b3["cnpj"] = (
    b3["cnpj"]
    .fillna(b3["cnpj_nome"])
)


b3.drop(
    columns=["cnpj_nome"],
    inplace=True
)


# ============================================================
# 10. IDENTIFICAR MÉTODO DO MATCH
# ============================================================

b3["match_method"] = "SEM_MATCH"


# CNPJ
b3.loc[
    b3["cnpj"].notna(),
    "match_method"
] = "CNPJ"


# Ticker
b3.loc[
    (
        b3["cnpj"].notna()
        &
        (b3["match_method"] == "SEM_MATCH")
    ),
    "match_method"
] = "TICKER"


# Nome
b3.loc[
    (
        b3["cnpj"].notna()
        &
        (b3["match_method"] == "SEM_MATCH")
    ),
    "match_method"
] = "NOME"


# ============================================================
# 11. FUZZY MATCH DE NOMES
# ============================================================

# Empresas únicas da CVM

nomes_cvm = (
    empresas[
        [
            "cnpj",
            "Nome_Companhia",
            "nome_norm"
        ]
    ]
    .drop_duplicates("cnpj")
)


# Apenas registros ainda sem CNPJ

sem_match = (
    b3["cnpj"].isna()
)


for idx in b3.index[sem_match]:

    nome_b3 = b3.at[idx, "nome_norm"]

    if not nome_b3:
        continue

    melhor_score = 0
    melhor_cnpj = None
    melhor_nome = None

    for _, empresa in nomes_cvm.iterrows():

        nome_cvm = empresa["nome_norm"]

        if not nome_cvm:
            continue

        score = SequenceMatcher(
            None,
            nome_b3,
            nome_cvm
        ).ratio()

        if score > melhor_score:

            melhor_score = score
            melhor_cnpj = empresa["cnpj"]
            melhor_nome = empresa["Nome_Companhia"]

    if melhor_score >= FUZZY_THRESHOLD:

        b3.at[idx, "cnpj"] = melhor_cnpj

        b3.at[idx, "match_method"] = (
            "FUZZY_NOME"
        )

        b3.at[idx, "match_score"] = (
            melhor_score
        )

        b3.at[idx, "Nome_Companhia_CVM"] = (
            melhor_nome
        )


# ============================================================
# 12. PREENCHER SCORE DOS MATCHES EXATOS
# ============================================================

b3["match_score"] = (
    b3["match_score"]
    if "match_score" in b3.columns
    else np.nan
)

b3.loc[
    b3["match_method"] == "CNPJ",
    "match_score"
] = 1.0

b3.loc[
    b3["match_method"] == "TICKER",
    "match_score"
] = 0.95

b3.loc[
    b3["match_method"] == "NOME",
    "match_score"
] = 0.90


# ============================================================
# 13. CONSTRUIR TABELA FINAL
# ============================================================

resultado = b3[
    [
        "isin",
        "cnpj",
        "ticker",
        "nome_empresa",
        "match_method",
        "match_score"
    ]
].copy()


# ============================================================
# 14. REMOVER ISIN VAZIO
# ============================================================

resultado = resultado[
    resultado["isin"] != ""
]


# ============================================================
# 15. REMOVER DUPLICATAS
# ============================================================

resultado = (
    resultado
    .drop_duplicates(
        subset=[
            "isin",
            "cnpj",
            "ticker"
        ]
    )
)


# ============================================================
# 16. ORDENAR
# ============================================================

resultado = (
    resultado
    .sort_values(
        [
            "ticker",
            "isin"
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# 17. SALVAR
# ============================================================

resultado.to_csv(
    ARQUIVO_SAIDA,
    index=False,
    encoding="utf-8-sig"
)


# ============================================================
# 18. RELATÓRIO
# ============================================================

print("\n")
print("=" * 80)
print("RESULTADO DO MATCH")
print("=" * 80)

print(
    f"Total de ISINs: "
    f"{resultado['isin'].nunique():,}"
)

print(
    f"Com CNPJ: "
    f"{resultado['cnpj'].notna().sum():,}"
)

print(
    f"Sem CNPJ: "
    f"{resultado['cnpj'].isna().sum():,}"
)

print("\nMétodo de matching:")

print(
    resultado["match_method"]
    .value_counts(dropna=False)
)

print("\nTaxa de cobertura:")

cobertura = (
    resultado["cnpj"].notna().mean()
)

print(
    f"{cobertura:.2%}"
)

print("\nArquivo gerado:")

print(ARQUIVO_SAIDA)


# ============================================================
# 19. MOSTRAR CASOS SEM MATCH
# ============================================================

sem_match_final = resultado[
    resultado["cnpj"].isna()
].copy()


print("\n")
print("=" * 80)
print("ISINs SEM CNPJ")
print("=" * 80)

print(
    sem_match_final.head(50)
)
```

```python
# ============================================================
# EXPORTAR LIQUIDEZ, VOLUME E MARKET CAP PARA O MODELO
# ============================================================
# Gera liquidez_mercado.csv no formato lido por
# momentum_etf/src/market_data.py:
#     date, ticker, volume_financeiro, market_cap
#
# - volume_financeiro: R$ negociados no dia (COTAHIST, mercado à vista).
# - market_cap: preço de fechamento da classe x quantidade de ações da
#   classe (ON para final 3, PN para finais 4-8), usando o capital social
#   mais recente informado à CVM ATÉ aquela data (merge_asof para trás,
#   sem look-ahead). Units (final 11) ficam sem market cap.
# - Só entram tickers de acoes_retornos.csv (df_wide), então units e
#   papéis como B3SA3 não são perdidos pelo regex [A-Z]{4}[3-8].
# - O mapeamento ticker -> CNPJ usa só os matches ALTO/MEDIO da célula
#   de matching (mapping_validado). Tickers sem match ficam com
#   market_cap vazio e o modelo, por padrão, não os filtra por tamanho.
#
# Copie o arquivo gerado para momentum_etf/data/.

ARQUIVO_LIQUIDEZ = "liquidez_mercado.csv"

tickers_modelo = set(df_wide.columns.astype(str).str.strip().str.upper())

# ------------------------------------------------------------
# 1. Volume e preço diários (mercado à vista, tickers do modelo)
# ------------------------------------------------------------

diario = df_b3[["data", "ticker", "tipo_mercado", "preco_ultimo", "volume_financeiro"]].copy()
diario["ticker"] = diario["ticker"].astype(str).str.strip().str.upper()
diario["tipo_mercado"] = diario["tipo_mercado"].astype(str).str.strip()
diario = diario[
    (diario["tipo_mercado"] == "010")
    & diario["ticker"].isin(tickers_modelo)
    & diario["data"].notna()
]

diario = (
    diario
    .sort_values(["ticker", "data"])
    .groupby(["data", "ticker"], as_index=False)
    .agg(
        volume_financeiro=("volume_financeiro", "sum"),
        preco_ultimo=("preco_ultimo", "last"),
    )
)

# ------------------------------------------------------------
# 2. Quantidade de ações por ticker ao longo do tempo
# ------------------------------------------------------------

cap = df_capital_social[
    [
        "CNPJ_Companhia",
        "Data_Referencia",
        "Quantidade_Acoes_Ordinarias",
        "Quantidade_Acoes_Preferenciais",
    ]
].copy()
cap["CNPJ"] = cap["CNPJ_Companhia"].astype(str).str.replace(r"\D", "", regex=True).str.zfill(14)
cap["Data_Referencia"] = pd.to_datetime(cap["Data_Referencia"], errors="coerce")
for col in ["Quantidade_Acoes_Ordinarias", "Quantidade_Acoes_Preferenciais"]:
    cap[col] = pd.to_numeric(cap[col], errors="coerce")
cap = cap.dropna(subset=["Data_Referencia"])

mapa = mapping_validado[["ticker", "CNPJ"]].copy()
mapa["ticker"] = mapa["ticker"].astype(str).str.strip().str.upper()
mapa["CNPJ"] = mapa["CNPJ"].astype(str).str.replace(r"\D", "", regex=True).str.zfill(14)
mapa = mapa.drop_duplicates("ticker")

acoes = mapa.merge(cap, on="CNPJ", how="inner")
acoes["quantidade_acoes"] = np.select(
    [acoes["ticker"].str.fullmatch(r"[A-Z0-9]{4}3"), acoes["ticker"].str.fullmatch(r"[A-Z0-9]{4}[4-8]")],
    [acoes["Quantidade_Acoes_Ordinarias"], acoes["Quantidade_Acoes_Preferenciais"]],
    default=np.nan,
)
acoes = (
    acoes[acoes["quantidade_acoes"] > 0]
    .sort_values(["ticker", "Data_Referencia"])
    .drop_duplicates(["ticker", "Data_Referencia"], keep="last")
    [["ticker", "Data_Referencia", "quantidade_acoes"]]
)

# ------------------------------------------------------------
# 3. Market cap diário = preço x última quantidade conhecida
# ------------------------------------------------------------

diario = pd.merge_asof(
    diario.sort_values("data"),
    acoes.sort_values("Data_Referencia"),
    left_on="data",
    right_on="Data_Referencia",
    by="ticker",
    direction="backward",
)
diario["market_cap"] = diario["preco_ultimo"] * diario["quantidade_acoes"]

liquidez_mercado = (
    diario
    .rename(columns={"data": "date"})
    [["date", "ticker", "volume_financeiro", "market_cap"]]
    .sort_values(["date", "ticker"])
    .reset_index(drop=True)
)

liquidez_mercado.to_csv(ARQUIVO_LIQUIDEZ, index=False)

# ------------------------------------------------------------
# 4. Diagnóstico de cobertura
# ------------------------------------------------------------

print("Arquivo salvo:", ARQUIVO_LIQUIDEZ, liquidez_mercado.shape)
print("Período:", liquidez_mercado["date"].min(), "até", liquidez_mercado["date"].max())
print(f"Tickers do modelo: {len(tickers_modelo):,}")
print(f"  com volume:     {liquidez_mercado['ticker'].nunique():,}")
print(f"  com market cap: {liquidez_mercado.dropna(subset=['market_cap'])['ticker'].nunique():,}")
print("\nTickers do modelo sem nenhum volume na B3:")
print(sorted(tickers_modelo - set(liquidez_mercado["ticker"]))[:50])
```
